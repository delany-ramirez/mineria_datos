# Ejemplo carga de archivos a Python con diferentes formatos


Introducción panorámica al proceso de ingestión de datos en pandas. Se muestra cómo un mismo conjunto de información puede residir en varios contenedores (texto plano, marcado, binario) y qué ajustes de lectura necesita cada uno para lograr un DataFrame coherente y bien tipado.

## Formato .csv

Texto plano delimitado por comas, ligero y casi universal: ideal para intercambios rápidos de datos tabulares. Requiere definir separador, codificación y, si hay comas embebidas, reglas de comillas y escape. Es el “lenguaje franca” entre hojas de cálculo, bases de datos y scripts.

```python
import pandas as pd

data_csv = pd.read_csv('data_load/mtcars.csv')

data_csv.head()
```

## Formato .tsv

Variante del CSV que usa tabulaciones como separador; minimiza choques con valores que ya incluyen comas. Muy apreciado en bioinformática y entornos UNIX. Basta indicar `sep='\t'` en `read_csv` — el resto se comporta igual que un CSV.

```python
data_tsv = pd.read_csv('data_load/mtcars.tsv', sep='\t')

data_tsv.head()
```

## Formato .json

Estructura jerárquica de pares clave-valor, perfecta para datos anidados o provenientes de API web. Pandas detecta listas y diccionarios y los “aplana” según corresponda; permite conservar la semántica original (objetos, arrays) que no cabe en una tabla CSV tradicional.

```python
data_json = pd.read_json('data_load/mtcars.json')

data_json.head()
```

## Formato .xml

Lenguaje de marcado con etiquetas auto-descriptivas, habitual en servicios antiguos y especificaciones de industria. Su jerarquía exige un parser (por ejemplo, lxml) y, al leerlo, es crucial apuntar a la profundidad (xpath) correcta para convertir nodos repetidos en filas de un DataFrame.

```python
# Requires installation of lxml library

data_xml = pd.read_xml('data_load/mtcars.xml')

data_xml.head()
```

## Formato .xlsx

Contenedor binario (en realidad un ZIP con XML interno) que soporta múltiples hojas, fórmulas y tipos ricos como fechas. Con `read_excel` y motores como *openpyxl* puedes traer una o varias hojas, elegir rangos y controlar las cabeceras. Es la mejor opción cuando el archivo proviene directamente de usuarios de Excel y se necesita mantener metadatos de celdas.

```python
# Requires installation of openpyxl library

data_excel = pd.read_excel('data_load/mtcars.xlsx')

data_excel.head()
```
